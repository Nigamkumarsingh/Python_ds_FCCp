# Project #1 web scraping with pandas 

```python
#reading a single csv file from website and renaming columns
import pandas as pd
df_data = pd.read_csv('https://www.football-data.co.uk/mmz4281/2526/E0.csv')
df_data.head()
df_data.rename(columns= {'FTHG':'home_goals','FTAG':'away_goals'},
               inplace=True)#updating the changes not copy
df_data.head()
```

```python
#reading multiple csv files from website
root = 'https://www.football-data.co.uk/mmz4281/2526/'#create root variable for all the links ie the part common in all the links
leagues = [ 'E0' , 'E1' , 'E2' , 'E3' , 'EC']#list where the difference lies in all the links
frames = []#empty list to store all the df
for league in leagues:#loop through the links to read it
    df = pd.read_csv( root +  league + ".csv",
                     encoding='unicode_escape')
    frames.append(df)
df.head()
```

```python
len(frames)#total how many links scanned
for i in range(5):
    print(frames[i].head())
```

```python
#above scanning was done only for one season but what if we wanted data for multiple seasons
#scanning for mulitple seasons
leagues = [ 'E0' , 'E1' , 'E2' , 'E3' , 'EC']
frames = []
root = 'https://www.football-data.co.uk/mmz4281/'
for league in leagues:
    for season in range(20,26):
        df = pd.read_csv( root + str(season) + str(season+1) +  '/' + league +'.csv',
                         encoding='unicode_escape')
        df['season']=str(season) + str(season+1)
        df['league']=league
        frames.append(df)
data = pd.concat(frames,ignore_index=True)
data.head()
```
