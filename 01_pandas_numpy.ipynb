```python
#adding a new column with assign
import numpy as np
import pandas as pd
df = pd.read_csv('StudentsPerformance.csv')
score_1 = np.random.randint(1,100,size = 1000)
score_2 = np.random.randint(1,100,size = 1000)
series_1 = pd.Series(score_1,index = np.arange(0,1000))
series_2 = pd.Series(score_2,index = np.arange(0,1000))
df = df.assign(score1 = series_1,score2 = series_2)
df
```

# operations on df

```python
import pandas as pd
df = pd.read_csv('StudentsPerformance.csv')
df.describe()
```

```python
#calculating on single column
print(df['math score'].sum())#calculating total of a particular column
print(df['math score'].count())
print(df['math score'].mean())
print(df['math score'].median())
print(df['math score'].mode())
print(df['math score'].std())
print(df['math score'].min())
print(df['math score'].max())
```

```python
#calculation on multiple columns
import pandas as pd
df = pd.read_csv('StudentsPerformance.csv')
df['math score'] + df['reading score'] + df['writing score']#summing multiple column values
df['average score']=(df['math score'] + df['reading score'] + df['writing score'])/3
df.head()
```

```python
df.head().round(2)
```

```python
#value counts
df['parental level of education'].value_counts(normalize=True)
len(df['average score'].value_counts(normalize=True).round(2))
```

```python
#sorting a df
import pandas as pd
df = pd.read_csv('StudentsPerformance.csv')
df.sort_values(by='math score')#by can be removed still same answer
df.sort_values('math score',ascending=False)
df.sort_values(['math score','reading score'],ascending=True).head()
```

```python
#update values in df
df.sort_values(['math score','reading score'],ascending=False,inplace=True)
```

```python
#sorting with text data
df.sort_values(['gender','parental level of education','race/ethnicity'],ascending=True,
               key= lambda col:col.str.lower()).head()
```

```python
#create index
import random
import numpy as np
import pandas as pd
df = pd.read_csv('StudentsPerformance.csv')
new_index = np.arange(0,1000)
random.shuffle(new_index)
df['new_index_col'] = new_index
df.head()
```
