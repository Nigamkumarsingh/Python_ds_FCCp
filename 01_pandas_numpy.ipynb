```python
#creating an data frame
#data frame is a 2D table like data structure made of rows and columns similar to excel sheet/sql table
data = np.array([[2,3],[3,1],[6,0]])
df = pd.DataFrame(data,index = ['row_1','row_2','row_3'],
                  columns=['col_1','col_2'])
df#index for rows columns for columns pass list for conversion
```

```python
#data frame using lists 
data = [[2,3],[43,2],[4,9]]
df = pd.DataFrame(data,index =[1,2,3],columns=[1,2])
df
```

```python
#data frame using dictionary
states = ['Assam','Bihar','New_york']
population = [78000,123000,340000]
dict_states = {'States':states,'Population':population}
df_population = pd.DataFrame(dict_states,index=['state_1','state_2','state_3'])
df_population
```

```python
df = pd.read_csv('StudentsPerformance.csv')
# df.info()#showing the info of the df
df.describe()#describing basic statistics of the df
print(len(df))#get length of the df/no of rows in df
print(max(df.index))#obtaiing the highest index of the df
print(min(df.index))#obtaining the minimum index of the df
print(df.dtypes)#obtaiing the data type of the columns
print(type(df))#obtaining the data type
round(df,2)#rounding the values of the dataset
round(df.describe(),2)#rounding of the basic statistics of the df upto 2 decimals 
```
