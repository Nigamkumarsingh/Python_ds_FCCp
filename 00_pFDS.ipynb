##Basic String methods

```python
print(("hello world!").lower())##converting to lower case
print(("hello world!").upper())##converting to upper case
print(("hello world!").title())##First letter capitalized
print(("hello world").count('d'))##count no of a particular letter
print(("hello world!").replace('l','v'))##replace a letter with different one
```

```python
#list -> store multiple items in a single container
#indexing from -1 means starting from the end of the list
#which means -1 index is of the last element of the list
countries = [ "India" , "Taiwan" , "Russia" , "Bhutan"]
length = len(countries)
for i in range(-1,-(length)-1,-1):
    print(countries[i])
```

```python
#list_slicing ***list_name[start:stop:step]***
#stop is not included in list slicing it will stop at the previous element
print(countries[0:3])
print(countries[-1:-4:-1])
#for negative indexing in list slicing remember that it works for left to right but if moving from right to left just give the the below frame
#list_name[start:end:step] **example list_name[-1:-3:-1]
print(countries[0:])#from first to last element in the list
print(countries[:2])#from first to a particular index
```

```python
#list_name.append("element") defaultly adding elements to the end of the list
countries = [ "India" , "Taiwan" , "Russia"]
countries.append("Spain")
#list_name.insert(index,"element") adding element to a specific index
countries.insert(2,"Africa")
countries_2 = ["Cambodia" , "New zeland"]
#meging two lists
print(countries + countries_2)
print(countries.append(countries_2))
nested_list = [countries , countries_2]#list inside a list
```

```python
#remvoing an element in the list
countries = [ "India" , "Taiwan" , "Russia" , "Bhutan"]
countries.remove("Bhutan")
print(countries)
countries.pop(2)
print(countries)
del countries[1]
print(countries)
```

```python
#sorting a list
num_list = [2,32,34,4,5,1,4,5,67]
num_list.sort()
print(num_list)
num_list.sort(reverse=True)
print(num_list)
#replacing a particular number with different one
num_list = [ 45 if x == 2 else x for x in num_list]
print(num_list)
num_list[0]=1000
print(num_list)
```

```python
#countries[:],countries.copy()
# copying a list 
countries = [ "India" , "Taiwan" , "Russia" , "Bhutan"]
new_c_list = countries[:]
print(new_c_list)
new_c_list_2 = countries.copy()
print(new_c_list_2)
```

```python
#Dictionary
my_dict = { 'key':'value','key2':'value2'}
my_data = {'name':'nigam', 'age':25, 'address':'chabua'}
print(my_data)
print(my_data.keys())#get the keys
print(my_data.values())#get the values
print(my_data.items())#get the pair values
my_data['pincode'] = 786184#add an item in the dictionary
print(my_data)
#update value in dictionary
my_data.update({'pincode':12345})
new_dictionary = my_data.copy()
print(new_dictionary)
### if changes made in the orginal dictionary it will not reflect in the copied dictionary
```

```python
##removing an item(key and value) in the dictionary
my_data = {'name':'nigam', 'age':25, 'address':'chabua'}
my_data.pop('age')#remove the mentioned key and value associated with it
print(my_data)
del my_data['address']#remove the mentioned key and value associated with it
print(my_data)
my_data.clear()#remove all elements in the dictionary//clear the dictionary
print(my_data)
```

```python
#functions
def sum_num( a, b):
    sum = (a+b)
    return sum
def multiply(a,b):
    product = a*b
    return product
print(sum_num(10,89.99))
print(multiply(3.6,4.5))
```

```python
#built in functions
countries = ['India' , 'Russia' , 'Taiwan']
for country in countries:
    print(len(country))#len calculate the length of item,list etc
```

```python
### some built in functions
num_list_1 = [ 10 , 45 , 53 , 34 , 5 , 99 ]
num_list_1.sort(reverse = True) #sorting and reversing the list
print(num_list_1)
print(type(num_list_1))## returns the type
print(len(num_list_1))#returns length of the list
print(max(num_list_1))#returns maximum
print(min(num_list_1))#returns minimum
num_list_1.sort()#sorting the list
print(num_list_1)
num_list_1.sort(reverse = False)#sorting and not reversing the list
print(num_list_1)
```
